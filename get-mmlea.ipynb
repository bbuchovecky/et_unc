# Multi-Model Large Ensemble Archives (MMLEA) on `glade`

```python
from pathlib import Path
import numpy as np
import xarray as xr
```

```python
LE_ROOTDIR = Path("/glade/campaign/collections/rda/data/d651039")

LE_NAMES = sorted([d.stem for d in LE_ROOTDIR.iterdir()])
LE_NAMES.remove("observations")
LE_NAMES.remove("cvdp")

print(LE_ROOTDIR)
print(LE_NAMES)
```

```python
table_id = "Amon"
variable_id = "evspsbl"

avail = []
not_avail = []

print(f"=== {variable_id}, {table_id} ===")
print("\nAVAILABLE")
for le in sorted(LE_NAMES):
    file_pattern = f"{variable_id}_{table_id}*historical*.nc"
    if (LE_ROOTDIR / le / table_id / variable_id).exists():
        files = list((LE_ROOTDIR / le / table_id / variable_id).glob(file_pattern))
        print(f"{le:16} {len(files)} members")
        avail.append(le)
    else:
        not_avail.append(le)

print("\nNOT AVAILABLE")
for le in not_avail:
    print(le)
```
